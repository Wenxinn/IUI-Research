4. find threshold that has the closest FRR and FAR
5. Take an average of FRR and FAR of the threshold in 4 for EER

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix

file_path = 'FRANK_STROKE_DATA_Normalized4.csv'
data = pd.read_csv(file_path)

if 'FRS_USER_ID' not in data.columns:
    raise ValueError("The 'FRS_USER_ID' column is missing from the dataset.")

users = list(map(int, data['FRS_USER_ID'].unique()))

EER_df = pd.DataFrame(index=users, columns=users)

feature_columns = [col for col in data.columns if col != 'FRS_USER_ID']
categorical_cols = data[feature_columns].select_dtypes(include=['object', 'category']).columns.tolist()
numeric_cols = data[feature_columns].select_dtypes(include=['number']).columns.tolist()

from sklearn.model_selection import train_test_split

TRAIN_SIZE = 120
TEST_SIZE = 60

pos_users = users
neg_users = users

for pos_user in pos_users:
    for neg_user in neg_users:
        if pos_user == neg_user: continue

        legitimate_user_data = data[data['FRS_USER_ID'] == pos_user]
        illegitimate_user_data = data[data['FRS_USER_ID'] == neg_user]
        other_user_data = data[(data['FRS_USER_ID'] != pos_user) & (data['FRS_USER_ID'] != neg_user)]

        legit_train, legit_test = train_test_split(
            legitimate_user_data, train_size=TRAIN_SIZE, test_size=TEST_SIZE, random_state=42
        )

        illegit_train = other_user_data.sample(
            n=TRAIN_SIZE, random_state=42, replace=False
        )

        illegit_test = illegitimate_user_data.sample(
            n=TEST_SIZE, random_state=42, replace=False
        )

        training_data = pd.concat([legit_train, illegit_train]).reset_index(drop=True)
        test_data = pd.concat([legit_test, illegit_test]).reset_index(drop=True)

        X_train = training_data.loc[:,'FRS_INTER_STROKE_TIME':]
        y_train = (training_data['FRS_USER_ID'] == pos_user).astype(int)

        X_test = test_data.loc[:,'FRS_INTER_STROKE_TIME':]
        y_test = (test_data['FRS_USER_ID'] == pos_user).astype(int)

        clf = RandomForestClassifier(random_state=42)
        try:
            clf.fit(X_train, y_train)
        except ValueError as e:
            print(f"Error training model for pair ({pos_user}, {neg_user}): {e}")
            EER_df.loc[pos_user, neg_user] = np.nan
            continue

        importances = clf.feature_importances_
        feature_names = X_train.columns
        feature_importance_df = pd.DataFrame({'feature': feature_names, 'importance': importances})
        feature_importance_df.sort_values(by='importance', ascending=False, inplace=True)

        feature_importance_df['cumulative_importance'] = feature_importance_df['importance'].cumsum()

        cutoff = 0.999
        selected_features = feature_importance_df[feature_importance_df['cumulative_importance'] <= cutoff]['feature']

        if selected_features.empty:
            selected_features = feature_importance_df['feature']

        X_train = X_train[selected_features]
        X_test = X_test[selected_features]

        clf = RandomForestClassifier(random_state=42)
        try:
            clf.fit(X_train, y_train)
        except ValueError as e:
            print(f"Error training model for pair ({pos_user}, {neg_user}) after feature selection: {e}")
            EER_df.loc[pos_user, neg_user] = np.nan
            continue

        y_pred = clf.predict_proba(X_test)[:, 1]

        thresholds = np.arange(0, 1.01, 0.01)
        frr_values = []
        far_values = []

        for threshold in thresholds:
            binary_predictions = [1 if score >= threshold else 0 for score in y_pred]

            cm = confusion_matrix(y_test, binary_predictions, labels=[0, 1])

            if cm.shape != (2, 2):
                EER_df.loc[pos_user, neg_user] = np.nan
                break

            TN, FP, FN, TP = cm.ravel()

            FRR = FN / (TP + FN) if (TP + FN) > 0 else 0
            FAR = FP / (FP + TN) if (FP + TN) > 0 else 0

            frr_values.append(FRR)
            far_values.append(FAR)
        else:
            frr_array = np.array(frr_values)
            far_array = np.array(far_values)
            abs_diffs = np.abs(frr_array - far_array)
            idxE = np.argmin(abs_diffs)
            EER = (frr_array[idxE] + far_array[idxE]) / 2

            EER_df.loc[pos_user, neg_user] = EER

        print(f'Pos: {pos_user}, Neg: {neg_user}   done')

EER_df.to_csv('EERs18.csv', index=True)
```

```python
import pandas as pd
import numpy as np
from scipy.stats import ttest_rel

def paired_t_test_all_values(csv_file_1, csv_file_2):

    df1 = pd.read_csv(csv_file_1)
    df2 = pd.read_csv(csv_file_2)

    df1_numeric = df1.apply(pd.to_numeric, errors='coerce')
    df2_numeric = df2.apply(pd.to_numeric, errors='coerce')

    data1 = df1_numeric.values.flatten()
    data2 = df2_numeric.values.flatten()

    valid_mask = ~np.isnan(data1) & ~np.isnan(data2)
    data1 = data1[valid_mask]
    data2 = data2[valid_mask]

    t_stat, p_value = ttest_rel(data1, data2)
    print(f"Paired t-test value: {t_stat}")
    print(f"p value: {p_value}")
    
if __name__ == "__main__":
    file_path_1 = "EERs.csv"
    file_path_2 = "EERs4.csv"
    paired_t_test_all_values(file_path_1, file_path_2)
```
